```python
import pandas as pd

df = pd.read_csv("../data_processed/master_test_dataset.csv")
```

```python
team1_df = df.copy()

team1_df["country"] = team1_df["team1"]
team1_df["opponent"] = team1_df["team2"]
```

```python
def get_result_team1(row):

    if pd.isna(row["winner"]):
        return "Draw"

    elif row["winner"] == row["team1"]:
        return "Win"

    else:
        return "Loss"

team1_df["result"] = team1_df.apply(get_result_team1, axis=1)
```

```python
def get_location_team1(row):

    if row["home_team"] == row["team1"]:
        return "Home"

    elif row["away_team"] == row["team1"]:
        return "Away"

    else:
        return "Neutral"

team1_df["location"] = team1_df.apply(get_location_team1, axis=1)
```

```python
team2_df = df.copy()

team2_df["country"] = team2_df["team2"]
team2_df["opponent"] = team2_df["team1"]
```

```python
def get_location_team2(row):

    if row["home_team"] == row["team2"]:
        return "Home"

    elif row["away_team"] == row["team2"]:
        return "Away"

    else:
        return "Neutral"

team2_df["location"] = team2_df.apply(get_location_team2, axis=1)
```

```python
final_df = pd.concat(
    [team1_df, team2_df],
    ignore_index=True
)
```
